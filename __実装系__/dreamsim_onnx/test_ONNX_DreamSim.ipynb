```python
import random
from pathlib import Path
from img_util import display_images_grid_html

target_dataset_dir = Path("../../sample_data/WIDER_OpenData/")

sample_img1 = random.choice(list(target_dataset_dir.glob("**/*.*")))
sample_img2 = random.choice(list(target_dataset_dir.glob("**/*.*")))

display_images_grid_html([sample_img1, sample_img2], ["sample_img1", "sample_img2"])
```

### PyTorch版

```python
from dreamsim import dreamsim
from PIL import Image

device = "cuda"
model, preprocess = dreamsim(
        dreamsim_type="dino_vitb16", 
        use_patch_model=False,
        cache_dir="./models/", 
        pretrained=True, 
        device=device
    )

img1 = preprocess(Image.open(sample_img1)).to(device)
img2 = preprocess(Image.open(sample_img2)).to(device)
distance = model(img1, img2)

print(f"Distance between images: {float(distance)}")
print(float(distance))
```

### ONNX版


```python
from dreamsim_onnx_utils import DreamSimONNX

# 1. ONNXモデルをロードする
model_path = "./onnx_models/dino_vitb16.onnx"
dreamsim_onnx = DreamSimONNX(model_path)

# 2. 2つの画像間の距離を計算
distance = dreamsim_onnx.get_distance(str(sample_img1), str(sample_img2))
print(f"Distance between images: {distance}")
```
